```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings # used to ignore mix datatype
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("cars.csv")
df
```

#### Check Duplicate Records

```python
df[df.duplicated]
```

```python
duplicated_rows = df[df.duplicated]
print("number of duplicated rows: ", duplicated_rows.shape)
```

#### Drop Duplicate Records

```python
df = df.drop_duplicates()
df
```

#### Check missing or null values

```python
print(df.isnull().sum())
```

#### Drop missing or null values

```python
df.dropna()
```

#### Show top 10 average MSRP (Price) per car brand (rounded figure)?

```python
df.groupby("Make")["MSRP"].mean().sort_values(ascending=False).head(10).round() # sorting values to get top 10
```

#### Show the average MSRP (price) per years (rounded figure)?

```python
avg_msrp = df.groupby('Year')['MSRP'].mean().round()
avg_msrp
```

#### Show average Engine HP (power) rounded figure and MPG (efficiency) rounded figure per year?

Hint: Average of (city mpg + highway MPG)

```python
df['MPG'] = (df["highway MPG"] + df["city mpg"])/2

avg_hp_mpg = df.groupby("Year")[["Engine HP","MPG"]].mean().round()
avg_hp_mpg
```

#### What percentage of cars are Compact, Midsize, or Large?

```python
vehicle_size = (df["Vehicle Size"].value_counts(normalize = True)*100).round(2).reset_index(name= "Percentage (%)")
vehicle_size
```

#### Which vehicle styles are the most sold (most common) in the dataset?

```python
style_count = df["Vehicle Style"].value_counts().reset_index(name= "Counts")
style_count
```

#### Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?

```python
cond = (df["MSRP"] >= 40000) & (df["Popularity"] >= 1000)

brands = df[cond]["Make"].value_counts().reset_index(name = "Vehicle Count")
brands
```
