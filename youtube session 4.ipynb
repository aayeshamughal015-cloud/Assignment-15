#### CONTROL FLOW STATEMENTS IN PYTHON 
##### What is a control flow statement?
From technical perspective, we can say that flow is nothing but many statements consisting of multiple operations which need to be executed.
Example : adding 10 numbers and finding the mean of the number.
We don't have any control over the statements yet. By control, mean the decision power. In cording we make decisions according to the situation, we follow required statamenets and skip irrelevent ones.
THERE ARE WE HAVE 3 THREE TYPES TO CONTROL THE STATEMENT IN PYTHON:
> Conditional Statements : 1.if
                           2.if-else
                           3.if-elif-else
                           4.nested if-else
 
> Transfer Statements : 1.break
                        2.continue
                        3.pass

> Iterative Statements : 1.for
                         2.while

#### if - elif - else statement:

```python
city = input("enter the city")

if city == 'karachi':
    print('do darya')
elif city == 'mirpurkhas':
    print('farm house')
elif city == 'hyderabad':
    print('roof top')
else:
    print("record not found")
```
