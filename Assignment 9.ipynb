```python
import csv
```

```python
def get_grade(percentage):
    percentage = float(percentage)
    if percentage > 80:
        return "A"
    elif percentage > 70:
        return "B"
    elif percentage > 60:
        return "C"
    elif percentage > 50:
        return "D"
    else:
        return "F"
```

```python
in_file = open('student data.csv', mode='r')
in_file
```

```python
reader = csv.reader(in_file)
reader
```

```python
data = list(reader)
for row in data:
    print(row)
```

```python
header = data[0]
students = data[1:] 
```

```python
header.append("Grade")
print(header)
```

```python
out_file = open('student_data_with_grade', mode='w')
writer = csv.writer(out_file)
```

```python
writer.writerow(header)
```

```python
for row in students:
    percentage = row[5] 
    grade = get_grade(percentage) 
    row.append(grade)
    writer.writerow(row) 
    print(row)
```

```python
in_file.close()
out_file.close()
```
