```python
class Employee:
    def __init__(
        self,
        name,
        number,
        address,
        salary,
        post,
        year_of_joining,
        education,
        year_of_experience,
    ):
        self.name = name
        self.number = number  
        self.address = address
        self.salary = int(salary)
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print(f"   Name                : {self.name}")
        print(f"   Contact Number      : {self.number}")
        print(f"   Address             : {self.address}")
        print(f"   Designation/Post    : {self.post}")
        print(f"   Current Salary      : ${self.salary:,.2f}")
        print(f"   Year of Joining     : {self.year_of_joining}")
        print(f"   Education           : {self.education}")
        print(f"   Experience (Years)  : {self.year_of_experience}")


    def update_year_of_experience(self, new_experience): 
        self.year_of_experience = new_experience
        print(f" Updated {self.name}'s experience to {self.year_of_experience} years.")

    def increase_salary(self, amount):
        if amount > 0:
            self.salary += amount
            print(f" Salary increased for {self.name} by ${amount:}. New Salary: ${self.salary:}")
        else:
            print("Increment amount must be positive.")



print("=== DEMONSTRATION WITH EMPLOYEE 101 ===\n")

# Create instance employee101
employee101 = Employee(
    name="Ayesha Khan",
    number="0300-1234567",
    address="Block 5, Clifton, Karachi",
    salary=85000,
    post="Data Analyst",
    year_of_joining=2021,
    education="BS Data Science",
    year_of_experience=3,
)

# Print four attributes individually
print("Printing 4 attributes individually:")
print(f"1. Name: {employee101.name}")
print(f"2. Post: {employee101.post}")
print(f"3. Salary: ${employee101.salary}")
print(f"4. Experience: {employee101.year_of_experience} years\n")

# Calling all methods on employee101
print("Calling describe_employee():")
employee101.describe_employee()

print("Performing Updates:")
employee101.increase_salary(15000)
employee101.update_year_of_experience(4)

print("\nUpdated details for employee101:")
employee101.describe_employee()



print("=== CREATING AND MANAGING 10 EMPLOYEES ===\n")

employees = [
    Employee(
        "Ali Raza",
        "0301-1111111",
        "Gulshan, Karachi",
        60000,
        "Software Engineer",
        2023,
        "BS Computer Science",
        1,
    ),
    Employee(
        "Sara Ahmed",
        "0302-2222222",
        "F-8, Islamabad",
        95000,
        "Project Manager",
        2019,
        "MS Management",
        5,
    ),
    Employee(
        "Usman Tariq",
        "0303-3333333",
        "Model Town, Lahore",
        120000,
        "Senior Developer",
        2017,
        "BS Software Engineering",
        7,
    ),
    Employee(
        "Fatima Malik",
        "0304-4444444",
        "Cantonment, Rawalpindi",
        70000,
        "HR Specialist",
        2022,
        "MBA HR",
        2,
    ),
    Employee(
        "Bilal Sheikh",
        "0305-5555555",
        "Jamshoro, Sindh",
        55000,
        "Junior Mathematician",
        2024,
        "BS Mathematics",
        0,
    ),
    Employee(
        "Zainab Bibi",
        "0306-6666666",
        "DHA Phase 6, Lahore",
        110000,
        "UI/UX Lead",
        2018,
        "BS Graphics & Design",
        6,
    ),
    Employee(
        "Hamza Noman",
        "0307-7777777",
        "Saddar, Hyderabad",
        65000,
        "QA Engineer",
        2022,
        "BS IT",
        2,
    ),
    Employee(
        "Sana Hassan",
        "0308-8888888",
        "Blue Area, Islamabad",
        130000,
        "DevOps Lead",
        2016,
        "MS Computer Science",
        8,
    ),
    Employee(
        "Omer Farooq",
        "0309-9999999",
        "University Road, Peshawar",
        50000,
        "Intern Engineer",
        2024,
        "BS Computer Engineering",
        0,
    ),
    Employee(
        "Hira Shah",
        "0310-0000000",
        "PECHS, Karachi",
        80000,
        "Marketing Specialist",
        2021,
        "BBA Marketing",
        3,
    ),
]

# 1. Call describe_employee() for each instance
print("--- Initial Profiles of All 10 Employees ---\n")
for emp in employees:
    emp.describe_employee()

# 2. Perform updates using the class methods
print("--- Updating Selected Employee Records ---\n")

# Update Bilal's salary and experience after performance review
employees[4].increase_salary(10000)
employees[4].update_year_of_experience(1)

# Annual raise for Senior Developer Usman
employees[2].increase_salary(20000)
employees[2].update_year_of_experience(8)

# Promotion raise for Intern Omer
employees[8].increase_salary(15000)

print("\n--- Displaying Updated Profiles for Selected Employees ---\n")
employees[4].describe_employee()
employees[2].describe_employee()
employees[8].describe_employee()
```

```python
def create_employee(
    name,
    number,
    address,
    salary,
    post,
    year_of_joining,
    education,
    year_of_experience,
):
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": int(salary),
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "year_of_experience": year_of_experience,
    }


def describe_employee(employee):
    print(f"   Name                : {employee['name']}")
    print(f"   Contact Number      : {employee['number']}")
    print(f"   Address             : {employee['address']}")
    print(f"   Designation/Post    : {employee['post']}")
    print(f"   Current Salary      : ${employee['salary']:}")
    print(f"   Year of Joining     : {employee['year_of_joining']}")
    print(f"   Education           : {employee['education']}")
    print(f"   Experience (Years)  : {employee['year_of_experience']}")
   


def update_year_of_experience(employee, new_experience):
    employee["year_of_experience"] = new_experience
    print( f" Updated {employee['name']}'s experience to {employee['year_of_experience']} years." )


def increase_salary(employee, amount):
    if amount > 0:
        employee["salary"] += amount
        print(f" Salary increased for {employee['name']} by ${amount:}. New Salary: ${employee['salary']:}")
    else:
        print("Increment amount must be positive.")


print("=== DEMONSTRATION WITH EMPLOYEE 101 (WITHOUT CLASSES) ===\n")

# Create dictionary for employee101
employee101 = create_employee(
    name="Ayesha Khan",
    number="0300-1234567",
    address="Block 5, Clifton, Karachi",
    salary=85000,
    post="Data Analyst",
    year_of_joining=2021,
    education="BS Data Science",
    year_of_experience=3,
)


# Print four attributes individually using dictionary keys
print("Printing 4 attributes individually:")
print(f"1. Name: {employee101['name']}")
print(f"2. Post: {employee101['post']}")
print(f"3. Salary: ${employee101['salary']}")
print(f"4. Experience: {employee101['year_of_experience']} years\n")


# Calling helper functions on employee101
print("Calling describe_employee():")
describe_employee(employee101)

print("Performing Updates:")
increase_salary(employee101, 15000)
update_year_of_experience(employee101, 4)

print("\nUpdated details for employee101:")
describe_employee(employee101)


print("=== CREATING AND MANAGING 10 EMPLOYEES ===\n")

employees = [
    create_employee(
        "Ali Raza",
        "0301-1111111",
        "Gulshan, Karachi",
        60000,
        "Software Engineer",
        2023,
        "BS Computer Science",
        1,
    ),
    create_employee(
        "Sara Ahmed",
        "0302-2222222",
        "F-8, Islamabad",
        95000,
        "Project Manager",
        2019,
        "MS Management",
        5,
    ),
    create_employee(
        "Usman Tariq",
        "0303-3333333",
        "Model Town, Lahore",
        120000,
        "Senior Developer",
        2017,
        "BS Software Engineering",
        7,
    ),
    create_employee(
        "Fatima Malik",
        "0304-4444444",
        "Cantonment, Rawalpindi",
        70000,
        "HR Specialist",
        2022,
        "MBA HR",
        2,
    ),
    create_employee(
        "Bilal Sheikh",
        "0305-5555555",
        "Jamshoro, Sindh",
        55000,
        "Junior Mathematician",
        2024,
        "BS Mathematics",
        0,
    ),
    create_employee(
        "Zainab Bibi",
        "0306-6666666",
        "DHA Phase 6, Lahore",
        110000,
        "UI/UX Lead",
        2018,
        "BS Graphics & Design",
        6,
    ),
    create_employee(
        "Hamza Noman",
        "0307-7777777",
        "Saddar, Hyderabad",
        65000,
        "QA Engineer",
        2022,
        "BS IT",
        2,
    ),
    create_employee(
        "Sana Hassan",
        "0308-8888888",
        "Blue Area, Islamabad",
        130000,
        "DevOps Lead",
        2016,
        "MS Computer Science",
        8,
    ),
    create_employee(
        "Omer Farooq",
        "0309-9999999",
        "University Road, Peshawar",
        50000,
        "Intern Engineer",
        2024,
        "BS Computer Engineering",
        0,
    ),
    create_employee(
        "Hira Shah",
        "0310-0000000",
        "PECHS, Karachi",
        80000,
        "Marketing Specialist",
        2021,
        "BBA Marketing",
        3,
    ),
]

# 1. Describe all 10 employees
print("--- Initial Profiles of All 10 Employees ---\n")
for emp in employees:
    describe_employee(emp)

# 2. Perform updates using the functions
print("--- Updating Selected Employee Records ---\n")

# Update Bilal (index 4)
increase_salary(employees[4], 10000)
update_year_of_experience(employees[4], 1)

# Update Usman (index 2)
increase_salary(employees[2], 20000)
update_year_of_experience(employees[2], 8)

# Update Omer (index 8)
increase_salary(employees[8], 15000)

print("\n--- Displaying Updated Profiles for Selected Employees ---\n")
describe_employee(employees[4])
describe_employee(employees[2])
describe_employee(employees[8])
```
