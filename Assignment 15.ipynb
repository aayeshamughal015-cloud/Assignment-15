```python
def sql_query(q):
    return pd.read_sql_query(q,conn)
```

#### 1. List all cities where the area is greater than 5000, along with their population.

```python
q = '''select c.name, c.population
       from cities c
       inner join facts f
       on 
       c.facts_id = f.id
       where f.area > 5000'''
sql_query(q)
```

#### 2.List the names of capital cities where the death rate is greater than 2% and the birth rate is less than 20%

```python
q = '''select c.name from cities c
       inner join facts f 
       on 
       c.facts_id = f.id
    where c.capital = 1 
    and f.death_rate > 2
    and f.birth_rate < 20'''

sql_query(q)
```

3. List the cities with a population growth rate between 1% and 5%, sorted in desc
by population growth.

```python
q = '''select c.name, f.population_growth from cities c
       inner join facts f
       on
       facts_id = f.id
       where f.population_growth between 1 and 5
       order by f.population_growth desc'''

sql_query(q)
```

4. Find countries that start with &quot;A&quot; and do not have any cities listed in the cities
table. The query should return the country name and city name. Using left join

```python
SELECT f.name AS country_name, c.name AS city_name
FROM facts f
LEFT JOIN cities c ON f.id = c.facts_id
WHERE f.name LIKE 'A%'
  AND c.name IS NULL;
```

```python
q = '''select f.name as country_name, c.name as city_name
       from facts f
       left join cities c
       on
       f.id = c.facts_id
       where
       f.name like "A%"
       and c.name is null'''

sql_query(q)
```

5. List all capital cities and their corresponding death rate using a LEFT JOIN.

```python
q = ''' select c.name as capital_city, f.death_rate
        from cities c
        left join facts f
        on 
        facts_id = f.id
        where
        c.capital = 1'''

sql_query(q)
```
